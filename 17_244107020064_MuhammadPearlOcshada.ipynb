# **BAGIAN 1**

- PyPREP

digunakan untuk implementasi Preprocessing Pipeline (PREP) standart.PyPREP biasa digunakan untuk mendeteksi dan menghapus saluran (channels) data yang rusak, menghilangkan artifak gangguan listrik (garis listrik), dan melakukan referencing data EEG secara otomatis agar sinyal lebih bersih sebelum dianalisis.
- Scipy

Spicy digunakan untuk berbagai kebutuhan manipulasi data tingkat tinggi yang terdapat modul untuk integrasi numerik, optimasi algorithma, aljabar linear, pemrosesan sinyal(seperti membuat filter frekuensi), dan analisis statistik
- wandb

berfungsi sebagai buku catatan digital yang mencatat metrik performa secara real-time (contoh loss dan accuracy), menyimpan arsitektur nodel, melacak hyperparameter, serta memvisualisasikan grafik pelatihan agar eksperimen lebih tersetruktur dan mudah direproduksi

- pyECG

pyECG dirancang khusus untuk mempermudah ekstraksi informasi dari sinyal jantung. pyECG digunakan untuk mendeteksi puncak gelombang jantung (seperti kompleks QRS), menghitung Variabilitas detak jantung(Heart Rate Variability/HRV) dan menyaring noise pada sinyal EKG mentah
